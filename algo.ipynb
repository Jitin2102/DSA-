## Bubble_Sort Algo

```python
%%writefile bubbleSort.c

#include<stdio.h>
void bubbleSort(int arr[], int n) {
    for (int i = 0; i < n-1; i++) {
        int swapped =0;
        for (int j = 0; j < n-i-1; j++) {
            if (arr[j] > arr[j+1]) {
                swapped = 1;
                int temp = arr[j];   // swap arr[j] and arr[j+1]
                arr[j] = arr[j+1];
                arr[j+1] = temp;
            }
        }
        if (swapped == 0) {
            break;
        }
    }
}
int main() {
    int arr[] = {64, 34, 25, 12, 22, 11, 90};
    int n = sizeof(arr)/sizeof(arr[0]);
    bubbleSort(arr, n);
    printf("Sorted array: \n");
    for (int i=0; i < n; i++)
        printf("%d ", arr[i]);
    return 0;
}
```

## Inserton_Sort Algo  
####  In recursive insertion sort : if array len is 0 or 1 means sorted already other wise sort first n-1 elements  then place at sorted position and repaet untill it becomes single element to be sorted 

####### void insertionSort(int arr[], int n) {
#######     int i,key,j;
#######     for(int i = 1;i < n;i++){
#######         key = arr[i];
#######         j = i-1;
#######         while(j >= 0 && arr[j] > key){
#######             arr[j+1] = arr[j];
#######             j--;
#######         }
#######         arr[j+1] = key;
#######     }
####### }

```python

%%writefile insertionSort.c
#include <stdio.h>

void RinsertionSort(int arr[], int n) {
    if (n <= 1) {
        return;
    }
    RinsertionSort(arr, n - 1);
    int key = arr[n - 1];
    int j = n - 2;
    while (j >= 0 && arr[j] > key) {
        arr[j + 1] = arr[j];
        j--;
    }
    arr[j + 1] = key;
}

int main() {
    int arr[] = {12, 11, 13, 5, 6};
    int n = sizeof(arr)/sizeof(arr[0]);
    RinsertionSort(arr, n);
    printf("Sorted array: \n");
    for(int i = 0;i < n;i++){
        printf("%d ", arr[i]);
    }
    return 0;
}
```

#### Finding second maximum

```python
%%writefile secondmax.c

#include<stdio.h>
int secondMax(int arr[],int n){
    int m2=arr[0],m1 = arr[1];
    for(int i = 2;i < n;i++){
        if(m1 < arr[i]){
            m2 = m1;
            m1 = arr[i];
        }
        else if(m1 > arr[i] && m2 != arr[i]) m2 = arr[i];
    }
    return m2;
}
int main(){
    int arr[] = {23,45,13,56,35,16,89,56,78,90};
    int n = sizeof(arr) / sizeof(arr[0]);
    int res = secondMax(arr,n);
    printf("Second maximum in array is :  %d",res);
    return 0;
}
```
